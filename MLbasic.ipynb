```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('C:/Users/Zain/OneDrive/Desktop/mirzajamalnasir750/Machine Learning/cats_dataset.csv')
```

```python
print(data)
df = data
```

```python
# Total unique cat breeds
print("Total Cat Breeds:", df['Breed'].nunique())
```

```python
print(df['Breed'].unique())
```

```python
# Breed counts
breed_counts = df['Breed'].value_counts()

breed_counts
```

```python
breed_counts.values
```

```python
# Bar graph
plt.figure(figsize=(12, 6))
plt.bar(breed_counts.index, breed_counts.values)
#Graph is actually completed but labels and title are missing so adding them below yet on this line of code
plt.xlabel("Cat Breed")
plt.ylabel("Number of Samples")
plt.title("Number of Samples for Each Cat Breed")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

```

```python
df['Gender'].value_counts().plot(
    kind='pie', autopct='%1.1f%%',
    figsize=(6, 6)
)
plt.ylabel('')
plt.title('Cat Gender Distribution')
plt.show()
```

```python
import seaborn as sns
sns.histplot(df['Age (Years)'], bins=15)
plt.title('Cat Age Distribution')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Age (Years)',
    y='Weight (kg)',
    hue='Gender'
)
plt.title('Age vs Weight')
plt.show()
```

```python
corr = df[
    ['Age (Years)', 'Weight (kg)']
].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.boxplot(
    data=df,
    x='Gender',
    y='Weight (kg)'
)
plt.title('Weight by Gender')
plt.show()
```

```python

plt.figure(figsize=(14, 10))

counts = df['Breed'].value_counts()

ax = sns.barplot(
    x=counts.values,
    y=counts.index,
    color='cornflowerblue'
)

for i, count in enumerate(counts.values):
    ax.text(count + 0.2, i, str(count), va='center')

plt.title('Cat Breed Sample Distribution')
plt.xlabel('Number of Samples')
plt.ylabel('Cat Breed')
plt.xlim(0, counts.max() * 1.15)
plt.tight_layout()
plt.show()
```
